```python
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

What was the most frequent movie duration in the 1990s? 
Save an approximate answer as an integer called duration (use 1990 as the decade's start year)

```python
#filtering out 1990's movies from original dataframe
movies_1990s=netflix_df[(netflix_df['release_year']>=1990) & (netflix_df['release_year']<2000)]
duration=int(netflix_df['duration'].mode().iloc[0])
```

```python
print(int(duration.iloc[0]))
```

A movie is considered short if it is less than 90 minutes.
Count the number of short action movies released in the 1990s and save this integer as short_movie_count

```python
#Subset of Action movies from 1990s dataframe
action_movies_df=movies_1990s[movies_1990s['genre']=='Action']

#Short Action movies are filtered 
movie_list_short=action_movies_df[action_movies_df['duration']<90]
#counting the number of short action movies
short_movie_count=movie_list_short['title'].count()
print(short_movie_count)
```
